### Importing the libraries

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

### Preparing the dataset

```python
# Dataset
X = np.array([500, 650, 720, 750, 780, 850, 880, 1050]).reshape(-1, 1)   # Area
y = np.array([12, 13.5, 16.5, 17, 17.2, 19, 21, 24])                          # Price
```

### Visualizing the plotted variables

```python
# -------------------------------
# Visualization 1: Plotted Variables (Scatter Plot)
# -------------------------------
plt.figure()
plt.scatter(X, y)
plt.xlabel("Area (sq ft)")
plt.ylabel("Price (Lakhs)")
plt.title("House Area vs Price")
plt.show()
```

### Training the model

```python
# Create and train model
model = LinearRegression()
model.fit(X, y)
```

### Displaying the learned parameters (slope and y-intercept)

```python
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```

### Visualizing the regression line

```python
# -------------------------------
# Visualization 2: Regression Line
# -------------------------------
plt.figure()
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel("Area (sq ft)")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression Line")
plt.show()
```

### Predicting price of a new house

```python
# Prediction
area = 1200
predicted_price = model.predict([[area]])
print("Predicted price for", area, "sq ft =", predicted_price[0], "Lakhs")
```
